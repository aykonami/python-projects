```python
"""
project: tower of hanoi
description: solves the tower of hanoi problem using recursion and records each move state
"""

# tower of hanoi solver
def hanoi_solver(n):

    # initial rods setup
    rods = [list(range(n, 0, -1)), [], []]

    moves = []

    # record current state of rods
    def record_state():
        moves.append(f"{rods[0]} {rods[1]} {rods[2]}")

    # recursive move function
    def move_disks(num_disks, source, target, auxiliary):

        if num_disks == 1:
            disk = rods[source].pop()
            rods[target].append(disk)
            record_state()
            return

        move_disks(num_disks - 1, source, auxiliary, target)

        disk = rods[source].pop()
        rods[target].append(disk)
        record_state()

        move_disks(num_disks - 1, auxiliary, target, source)

    # initial state
    record_state()

    # solve puzzle
    move_disks(n, 0, 2, 1)

    return "\n".join(moves)
```
