```python
"""
lab: bisection method
description: approximates the square root of a number using the bisection method
"""

# calculate square root using the bisection method
def square_root_bisection(number, tolerance=0.01, max_iterations=100):

    if number < 0:
        raise ValueError("Square root of negative number is not defined in real numbers")

    if number == 0 or number == 1:
        print(f"The square root of {number} is {number}")
        return number

    low = 0
    high = max(1, number)

    for _ in range(max_iterations):

        mid = (low + high) / 2
        square = mid * mid

        if square < number:
            low = mid
        else:
            high = mid

        if (high - low) <= tolerance:
            root = mid
            print(f"The square root of {number} is approximately {root}")
            return root

    print(f"Failed to converge within 10 iterations")

    return None
```
