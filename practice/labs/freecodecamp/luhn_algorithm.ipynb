```python
"""
project: luhn algorithm
description: validates a card number using the luhn checksum algorithm
"""

# verify card number
def verify_card_number(card_number):

    # remove separators
    card_number = card_number.replace("-", "").replace(" ", "")

    total = 0
    reversed_digits = card_number[::-1]

    # process digits
    for index, digit in enumerate(reversed_digits):

        num = int(digit)

        # double every second digit
        if index % 2 == 1:

            num *= 2

            if num > 9:
                num -= 9

        total += num

    # validate checksum
    if total % 10 == 0:
        return "VALID!"

    return "INVALID!"
```
