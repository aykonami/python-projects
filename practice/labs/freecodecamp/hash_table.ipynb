```python
"""
lab: hash table
description: implements a simple hash table with add, remove, and lookup operations
"""

# hash table class
class HashTable:

    def __init__(self):
        self.collection = {}

    # generate hash value
    def hash(self, key):
        return sum(ord(char) for char in key)

    # add key-value pair
    def add(self, key, value):
        hashed_key = self.hash(key)

        if hashed_key not in self.collection:
            self.collection[hashed_key] = {}

        self.collection[hashed_key][key] = value

    # remove key-value pair
    def remove(self, key):
        hashed_key = self.hash(key)

        if hashed_key in self.collection and key in self.collection[hashed_key]:
            del self.collection[hashed_key][key]

            if not self.collection[hashed_key]:
                del self.collection[hashed_key]

    # find value by key
    def lookup(self, key):
        hashed_key = self.hash(key)

        if hashed_key in self.collection:
            return self.collection[hashed_key].get(key)

        return None
```
