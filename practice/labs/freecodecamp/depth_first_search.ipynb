```python
"""
project: depth first search
description: traverses a graph using the depth-first search algorithm
"""

# depth-first search function
def dfs(adj_matrix, start_node):

    visited = []
    stack = [start_node]

    while stack:

        node = stack.pop()

        if node not in visited:
            visited.append(node)

            # add neighbors to stack
            for neighbor in range(len(adj_matrix[node]) - 1, -1, -1):

                if adj_matrix[node][neighbor] == 1 and neighbor not in visited:
                    stack.append(neighbor)

    return visited
```
