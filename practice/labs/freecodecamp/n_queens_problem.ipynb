```python
"""
lab: n queens problem
description: solves the n queens problem using depth first search and backtracking
"""

# n queens solver
def dfs_n_queens(n):

    if n < 1:
        return []

    solutions = []

    # check if queen placement is safe
    def is_safe(board, row, col):

        for r in range(row):
            c = board[r]

            # same column
            if c == col:
                return False

            # same diagonal
            if abs(c - col) == abs(r - row):
                return False

        return True

    # depth-first search
    def dfs(row, board):

        if row == n:
            solutions.append(board[:])
            return

        for col in range(n):

            if is_safe(board, row, col):
                board.append(col)
                dfs(row + 1, board)
                board.pop()

    dfs(0, [])

    return solutions
```
