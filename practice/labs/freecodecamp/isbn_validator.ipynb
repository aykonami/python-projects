```python
"""
lab: isbn validator
description: validates isbn-10 and isbn-13 codes using check digit calculations
"""

# calculate isbn-10 check digit
def calculate_check_digit_10(isbn):
    total = 0

    for i in range(9):
        total += int(isbn[i]) * (10 - i)

    remainder = total % 11
    check = 11 - remainder

    if check == 10:
        return "X"
    elif check == 11:
        return "0"
    else:
        return str(check)

# calculate isbn-13 check digit
def calculate_check_digit_13(isbn):
    total = 0

    for i in range(12):
        if i % 2 == 0:
            total += int(isbn[i])
        else:
            total += int(isbn[i]) * 3

    remainder = total % 10
    check = 10 - remainder

    if check == 10:
        return "0"

    return str(check)

# validate isbn code
def validate_isbn(isbn, length):

    try:
        length = int(length)
    except ValueError:
        print("Length must be a number.")
        return

    if length not in (10, 13):
        print("Length should be 10 or 13.")
        return

    if length == 10 and len(isbn) != 10:
        print("ISBN-10 code should be 10 digits long.")
        return

    if length == 13 and len(isbn) != 13:
        print("ISBN-13 code should be 13 digits long.")
        return

    try:
        if length == 10:
            expected = calculate_check_digit_10(isbn)
        else:
            expected = calculate_check_digit_13(isbn)

        if isbn[-1].upper() == expected:
            print("Valid ISBN Code.")
        else:
            print("Invalid ISBN Code.")

    except ValueError:
        print("Invalid character was found.")
        return
    except IndexError:
        print("Invalid character was found.")
        return
    except TypeError:
        print("Invalid ISBN Code.")
        return

# run program
def main():

    try:
        user_input = input("Enter ISBN and length: ")
        isbn, length = user_input.split(",")
        validate_isbn(isbn, length)

    except ValueError:
        print("Enter comma-separated values.")
    except IndexError:
        print("Enter comma-separated values.")
```
