```python
"""
project: number pattern generator
description: generates a sequence of numbers from 1 to n as a formatted string
"""

# generate number pattern
def number_pattern(n):

    # validate input type
    if not isinstance(n, int):
        return "Argument must be an integer value."

    # validate range
    if n < 1:
        return "Argument must be an integer greater than 0."

    result = ""

    # build pattern
    for i in range(1, n + 1):
        result += str(i)

        if i != n:
            result += " "

    return result
```
