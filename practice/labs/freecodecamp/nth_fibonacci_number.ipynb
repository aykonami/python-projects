```python
"""
lab: nth fibonacci number
description: calculates the nth fibonacci number using an iterative approach
"""

# fibonacci function
def fibonacci(n):

    # starting sequence
    sequence = [0, 1]

    # handle first two numbers
    if n < 2:
        return sequence[n]

    # generate sequence
    for i in range(2, n + 1):
        sequence.append(sequence[i - 1] + sequence[i - 2])

    return sequence[n]
```
