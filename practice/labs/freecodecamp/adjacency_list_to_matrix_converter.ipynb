```python
"""
project: adjacency list to matrix converter
description: converts a graph from an adjacency list representation to an adjacency matrix
"""

# convert adjacency list to matrix
def adjacency_list_to_matrix(adj_list):

    n = len(adj_list)

    # create empty matrix
    matrix = [
        [0 for _ in range(n)]
        for _ in range(n)
    ]

    # populate matrix
    for node, neighbors in adj_list.items():

        for neighbor in neighbors:
            matrix[node][neighbor] = 1

    # display matrix
    for row in matrix:
        print(row)

    return matrix
```
