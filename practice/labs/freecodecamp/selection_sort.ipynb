```python
"""
project: selection sort
description: sorts a list using the selection sort algorithm
"""

# selection sort function
def selection_sort(items):

    n = len(items)

    # iterate through the list
    for i in range(n):

        # assume current position contains the minimum value
        min_index = i

        # find the smallest value in the remaining list
        for j in range(i + 1, n):
            if items[j] < items[min_index]:
                min_index = j

        # swap values if needed
        if min_index != i:
            items[i], items[min_index] = (
                items[min_index],
                items[i]
            )

    return items
```
