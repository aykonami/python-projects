```python
"""
project: breadth first search
description: generates valid parentheses combinations using breadth first search
"""

# generate valid parentheses
def gen_parentheses(pairs):

    # validate input
    if not isinstance(pairs, int):
        return 'The number of pairs should be an integer'

    if pairs < 1:
        return 'The number of pairs should be at least 1'

    # initialize queue and results
    queue = [('', 0, 0)]
    result = []

    # process queue
    while queue:

        current, opens_used, closes_used = queue.pop(0)

        # add completed combination
        if len(current) == 2 * pairs:
            result.append(current)

        else:

            # add opening parenthesis
            if opens_used < pairs:
                queue.append((current + '(', opens_used + 1, closes_used))

            # add closing parenthesis
            if closes_used < opens_used:
                queue.append((current + ')', opens_used, closes_used + 1))

    return result

# test cases
print(gen_parentheses(2))
print(gen_parentheses(3))
```
